```python

import os
import json
import random
import io
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from pathlib import Path
from PIL import Image, ImageFile

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from tensorflow.keras import layers, models, callbacks
from tensorflow.keras.applications import EfficientNetB0

warnings.filterwarnings("ignore")
ImageFile.LOAD_TRUNCATED_IMAGES = True


# ============================================================
# 1. KONFIGURASI
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

IMG_SIZE = (224, 224)
BATCH_SIZE = 16

EPOCHS_HEAD = 20
EPOCHS_FINE = 15

LEARNING_RATE_HEAD = 0.001
LEARNING_RATE_FINE = 0.00001

TEST_SIZE = 0.30
VALIDATION_SIZE = 0.20

# Empat kelas rempah
CLASS_NAMES = [
    "Jahe",
    "Kunyit",
    "Kencur",
    "Lengkuas"
]

NUM_CLASSES = len(CLASS_NAMES)

OUTPUT_DIR = Path("hasil_rimpang")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("TensorFlow:", tf.__version__)
print("GPU tersedia:", tf.config.list_physical_devices("GPU"))


# ============================================================
# 2. MENENTUKAN LOKASI DATASET
# ============================================================

# Jika folder tidak ditemukan otomatis,
# masukkan lokasi folder secara manual.
# Contoh: DATASET_DIR = Path(r"D:\Dataset rimpang")

DATASET_DIR = None

if DATASET_DIR is None:

    kandidat = [
        Path.cwd() / "Dataset rimpang",
        Path.cwd().parent / "Dataset rimpang"
    ]

    for folder in kandidat:

        if folder.exists() and folder.is_dir():
            DATASET_DIR = folder
            break

if DATASET_DIR is None or not DATASET_DIR.exists():

    raise FileNotFoundError(
        "Folder Dataset rimpang tidak ditemukan. "
        "Silakan tentukan DATASET_DIR secara manual."
    )

print("\nLokasi dataset:", DATASET_DIR.resolve())


# ============================================================
# 3. MEMBACA DAN MEMERIKSA DATASET
# ============================================================

data = []

EXTENSIONS = [
    ".jpg", ".jpeg", ".png",
    ".bmp", ".webp", ".tif", ".tiff"
]

jumlah_valid = 0
jumlah_rusak = 0

for class_name in CLASS_NAMES:

    class_folder = DATASET_DIR / class_name

    if not class_folder.exists():

        raise FileNotFoundError(
            f"Folder kelas tidak ditemukan: {class_folder}"
        )

    label = CLASS_NAMES.index(class_name)

    image_files = [
        p for p in class_folder.rglob("*")
        if p.is_file() and p.suffix.lower() in EXTENSIONS
    ]

    for image_path in image_files:

        try:

            with Image.open(image_path) as img:
                img.verify()

            data.append({

                "path": str(image_path),
                "label": label,
                "class_name": class_name

            })

            jumlah_valid += 1

        except Exception as e:

            print("\nGambar dilewati:", image_path)
            print("Penyebab:", e)

            jumlah_rusak += 1

df = pd.DataFrame(data)

if len(df) == 0:

    raise ValueError(
        "Tidak ada gambar valid yang ditemukan."
    )

print("\n===================================")
print("HASIL PEMERIKSAAN DATASET")
print("===================================")

print("Jumlah gambar valid :", jumlah_valid)
print("Jumlah gambar rusak :", jumlah_rusak)
print("Total gambar        :", len(df))

print("\nDistribusi setiap kelas:")

print(
    df["class_name"].value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)

if (df["class_name"].value_counts() == 0).any():

    raise ValueError(
        "Ada kelas yang tidak memiliki gambar."
    )


# ============================================================
# 4. PEMBAGIAN DATASET
# ============================================================

jumlah_per_kelas = df["class_name"].value_counts()

print("\nJumlah gambar setiap kelas:")
print(jumlah_per_kelas)

if jumlah_per_kelas.min() < 5:

    raise ValueError(
        "Ada kelas yang memiliki kurang dari 5 gambar. "
        "Tambahkan gambar agar pembagian stratified "
        "dapat dilakukan."
    )

# 70% data pengembangan dan 30% data testing

df_development, df_test = train_test_split(

    df,

    test_size=TEST_SIZE,

    random_state=SEED,

    stratify=df["label"]

)

# 20% dari data pengembangan menjadi validation

df_train, df_val = train_test_split(

    df_development,

    test_size=VALIDATION_SIZE,

    random_state=SEED,

    stratify=df_development["label"]

)

df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

print("\n===================================")
print("HASIL PEMBAGIAN DATASET")
print("===================================")

print("Total dataset :", len(df))
print("Training      :", len(df_train))
print("Validation    :", len(df_val))
print("Testing       :", len(df_test))

print(f"\nTraining   : {len(df_train)/len(df)*100:.2f}%")
print(f"Validation : {len(df_val)/len(df)*100:.2f}%")
print(f"Testing    : {len(df_test)/len(df)*100:.2f}%")

print("\nDistribusi kelas pada setiap dataset:")

print("\nTRAINING")
print(
    df_train["class_name"]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)

print("\nVALIDATION")
print(
    df_val["class_name"]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)

print("\nTESTING")
print(
    df_test["class_name"]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
)


# ============================================================
# 5. FUNGSI MEMBACA GAMBAR RGB
# ============================================================

# Semua gambar dikonversi menjadi RGB.
# Mengatasi gambar grayscale, RGBA, CMYK,
# dan format warna lainnya.

def decode_image_pil(image_bytes):

    try:

        image = Image.open(io.BytesIO(image_bytes))

        image.seek(0)

        image = image.convert("RGB")

        image = image.resize(

            (IMG_SIZE[1], IMG_SIZE[0]),

            Image.Resampling.BILINEAR

        )

        image = np.asarray(
            image,
            dtype=np.float32
        )

        return image

    except Exception as e:

        raise ValueError(
            f"Gagal membaca gambar: {e}"
        )


def load_image(path, label):

    image_bytes = tf.io.read_file(path)

    image = tf.numpy_function(

        func=decode_image_pil,

        inp=[image_bytes],

        Tout=tf.float32

    )

    image.set_shape([

        IMG_SIZE[0],
        IMG_SIZE[1],
        3

    ])

    label = tf.cast(label, tf.int32)

    return image, label


# ============================================================
# 6. MEMBUAT DATASET TENSORFLOW
# ============================================================

def create_dataset(dataframe, training=False):

    paths = dataframe["path"].values

    labels = dataframe["label"].values.astype(np.int32)

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    if training:

        dataset = dataset.shuffle(

            buffer_size=len(dataframe),

            seed=SEED,

            reshuffle_each_iteration=True

        )

    dataset = dataset.map(

        load_image,

        num_parallel_calls=tf.data.AUTOTUNE

    )

    dataset = dataset.batch(BATCH_SIZE)

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


train_ds = create_dataset(
    df_train,
    training=True
)

val_ds = create_dataset(df_val)

test_ds = create_dataset(df_test)

print("\nPipeline dataset berhasil dibuat.")


# ============================================================
# 7. VISUALISASI DATASET
# ============================================================

plt.figure(figsize=(12, 8))

sample_data = df.sample(

    n=min(12, len(df)),

    random_state=SEED

).reset_index(drop=True)

for i, row in sample_data.iterrows():

    image = decode_image_pil(

        Path(row["path"]).read_bytes()

    )

    plt.subplot(3, 4, i + 1)

    plt.imshow(image.astype("uint8"))

    plt.title(row["class_name"])

    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 8. CLASS WEIGHT
# ============================================================

classes = np.arange(NUM_CLASSES)

weights = compute_class_weight(

    class_weight="balanced",

    classes=classes,

    y=df_train["label"].values

)

class_weights = {

    int(cls): float(weight)

    for cls, weight in zip(classes, weights)

}

print("\nClass weight:")

for cls, weight in class_weights.items():

    print(
        f"{CLASS_NAMES[cls]}: {weight:.4f}"
    )


# ============================================================
# 9. DATA AUGMENTATION
# ============================================================

data_augmentation = models.Sequential([

    layers.RandomFlip("horizontal"),

    layers.RandomRotation(0.05),

    layers.RandomZoom(0.10),

    layers.RandomTranslation(

        height_factor=0.05,

        width_factor=0.05

    )

], name="data_augmentation")


# ============================================================
# 10. MEMBANGUN MODEL EFFICIENTNETB0
# ============================================================

base_model = EfficientNetB0(

    include_top=False,

    weights="imagenet",

    input_shape=(

        IMG_SIZE[0],

        IMG_SIZE[1],

        3

    )

)

base_model.trainable = False

inputs = layers.Input(

    shape=(

        IMG_SIZE[0],

        IMG_SIZE[1],

        3

    )

)

x = data_augmentation(inputs)

# EfficientNetB0 memiliki preprocessing internal.
# Input tetap menggunakan rentang 0-255.
# Jangan membagi gambar dengan 255.

x = base_model(

    x,

    training=False

)

x = layers.GlobalAveragePooling2D()(x)

x = layers.BatchNormalization()(x)

x = layers.Dense(

    128,

    activation="relu"

)(x)

x = layers.Dropout(0.5)(x)

# Output 4 kelas menggunakan Softmax

outputs = layers.Dense(

    NUM_CLASSES,

    activation="softmax"

)(x)

model = models.Model(

    inputs=inputs,

    outputs=outputs

)

model.summary()


# ============================================================
# 11. COMPILE MODEL TAHAP PERTAMA
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=LEARNING_RATE_HEAD

    ),

    loss="sparse_categorical_crossentropy",

    metrics=[

        "accuracy",

        tf.keras.metrics.SparseTopKCategoricalAccuracy(
            k=2,
            name="top2_accuracy"
        )

    ]

)


# ============================================================
# 12. CALLBACK TAHAP PERTAMA
# ============================================================

checkpoint_head = callbacks.ModelCheckpoint(

    filepath=str(
        OUTPUT_DIR / "best_head.keras"
    ),

    monitor="val_loss",

    save_best_only=True,

    mode="min",

    verbose=1

)

earlystop_head = callbacks.EarlyStopping(

    monitor="val_loss",

    patience=5,

    restore_best_weights=True,

    verbose=1

)

reduce_lr_head = callbacks.ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.2,

    patience=3,

    min_lr=0.000001,

    verbose=1

)


# ============================================================
# 13. TRAINING TAHAP PERTAMA
# ============================================================

print("\n===================================")
print("TRAINING TAHAP 1: FEATURE EXTRACTION")
print("===================================")

history_head = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=EPOCHS_HEAD,

    class_weight=class_weights,

    callbacks=[

        checkpoint_head,

        earlystop_head,

        reduce_lr_head

    ],

    verbose=1

)


# ============================================================
# 14. MEMUAT MODEL TERBAIK TAHAP PERTAMA
# ============================================================

model = tf.keras.models.load_model(

    str(OUTPUT_DIR / "best_head.keras")

)

print(
    "\nModel terbaik tahap pertama berhasil dimuat."
)


# ============================================================
# 15. FINE-TUNING EFFICIENTNETB0
# ============================================================

base_model = model.get_layer("efficientnetb0")

base_model.trainable = True

# Membekukan seluruh layer terlebih dahulu

for layer in base_model.layers:

    layer.trainable = False

# Membuka 30 layer terakhir
# Kecuali BatchNormalization

for layer in base_model.layers[-30:]:

    if not isinstance(

        layer,

        layers.BatchNormalization

    ):

        layer.trainable = True


print("\nJumlah layer EfficientNetB0:",
      len(base_model.layers))

print("Jumlah layer yang dapat dilatih:",
      sum(

          1 for layer in base_model.layers

          if layer.trainable

      ))


# ============================================================
# 16. COMPILE ULANG UNTUK FINE-TUNING
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=LEARNING_RATE_FINE

    ),

    loss="sparse_categorical_crossentropy",

    metrics=[

        "accuracy",

        tf.keras.metrics.SparseTopKCategoricalAccuracy(
            k=2,
            name="top2_accuracy"
        )

    ]

)


# ============================================================
# 17. CALLBACK FINE-TUNING
# ============================================================

checkpoint_fine = callbacks.ModelCheckpoint(

    filepath=str(
        OUTPUT_DIR / "best_fine.keras"
    ),

    monitor="val_loss",

    save_best_only=True,

    mode="min",

    verbose=1

)

earlystop_fine = callbacks.EarlyStopping(

    monitor="val_loss",

    patience=7,

    restore_best_weights=True,

    verbose=1

)

reduce_lr_fine = callbacks.ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.2,

    patience=3,

    min_lr=0.0000001,

    verbose=1

)


# ============================================================
# 18. TRAINING TAHAP KEDUA
# ============================================================

print("\n===================================")
print("TRAINING TAHAP 2: FINE-TUNING")
print("===================================")

history_fine = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=EPOCHS_FINE,

    class_weight=class_weights,

    callbacks=[

        checkpoint_fine,

        earlystop_fine,

        reduce_lr_fine

    ],

    verbose=1

)


# ============================================================
# 19. MEMILIH MODEL TERBAIK
# ============================================================

best_head_loss = min(
    history_head.history["val_loss"]
)

best_fine_loss = min(
    history_fine.history["val_loss"]
)

print("\nValidation loss terbaik tahap 1:",
      best_head_loss)

print("Validation loss terbaik tahap 2:",
      best_fine_loss)

if best_fine_loss < best_head_loss:

    print("\nModel terbaik: Fine-tuning")

    model = tf.keras.models.load_model(

        str(OUTPUT_DIR / "best_fine.keras")

    )

else:

    print("\nModel terbaik: Feature Extraction")

    model = tf.keras.models.load_model(

        str(OUTPUT_DIR / "best_head.keras")

    )


# Menyimpan model akhir

model.save(

    str(OUTPUT_DIR / "model_rimpang.keras")

)

# Menyimpan nama kelas

with open(

    OUTPUT_DIR / "class_names.json",

    "w"

) as f:

    json.dump(CLASS_NAMES, f, indent=4)

print("\nModel akhir berhasil disimpan.")


# ============================================================
# 20. GRAFIK ACCURACY DAN LOSS
# ============================================================

acc = (

    history_head.history["accuracy"]

    + history_fine.history["accuracy"]

)

val_acc = (

    history_head.history["val_accuracy"]

    + history_fine.history["val_accuracy"]

)

loss = (

    history_head.history["loss"]

    + history_fine.history["loss"]

)

val_loss = (

    history_head.history["val_loss"]

    + history_fine.history["val_loss"]

)

epochs_range = range(1, len(acc) + 1)

# Grafik accuracy

plt.figure(figsize=(12, 5))

plt.plot(

    epochs_range,

    acc,

    label="Training Accuracy"

)

plt.plot(

    epochs_range,

    val_acc,

    label="Validation Accuracy"

)

plt.axvline(

    x=len(history_head.history["accuracy"]),

    linestyle="--",

    label="Mulai Fine-tuning"

)

plt.title("Training and Validation Accuracy")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid()

plt.show()


# Grafik loss

plt.figure(figsize=(12, 5))

plt.plot(

    epochs_range,

    loss,

    label="Training Loss"

)

plt.plot(

    epochs_range,

    val_loss,

    label="Validation Loss"

)

plt.axvline(

    x=len(history_head.history["loss"]),

    linestyle="--",

    label="Mulai Fine-tuning"

)

plt.title("Training and Validation Loss")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 21. EVALUASI DATA TESTING
# ============================================================

print("\n===================================")
print("EVALUASI MODEL PADA DATA TESTING")
print("===================================")

test_results = model.evaluate(

    test_ds,

    verbose=1,

    return_dict=True

)

for name, value in test_results.items():

    print(f"{name}: {value:.4f}")


# ============================================================
# 22. CLASSIFICATION REPORT
# ============================================================

y_test_true = df_test["label"].values

y_test_prob = model.predict(

    test_ds,

    verbose=1

)

# Mengambil kelas dengan probabilitas tertinggi

y_test_pred = np.argmax(

    y_test_prob,

    axis=1

)

print("\n===================================")
print("CLASSIFICATION REPORT")
print("===================================")

print(

    classification_report(

        y_test_true,

        y_test_pred,

        labels=list(range(NUM_CLASSES)),

        target_names=CLASS_NAMES,

        zero_division=0,

        digits=4

    )

)


# ============================================================
# 23. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test_true,

    y_test_pred,

    labels=list(range(NUM_CLASSES))

)

disp = ConfusionMatrixDisplay(

    confusion_matrix=cm,

    display_labels=CLASS_NAMES

)

fig, ax = plt.subplots(figsize=(8, 7))

disp.plot(

    cmap="Blues",

    values_format="d",

    ax=ax

)

plt.title("Confusion Matrix - EfficientNetB0")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()


# ============================================================
# 24. METRIK EVALUASI
# ============================================================

accuracy = accuracy_score(

    y_test_true,

    y_test_pred

)

precision = precision_score(

    y_test_true,

    y_test_pred,

    average="macro",

    zero_division=0

)

recall = recall_score(

    y_test_true,

    y_test_pred,

    average="macro",

    zero_division=0

)

f1 = f1_score(

    y_test_true,

    y_test_pred,

    average="macro",

    zero_division=0

)

print("\n===================================")
print("HASIL EVALUASI MODEL")
print("===================================")

print(f"Accuracy          : {accuracy:.4f}")

print(f"Precision (Macro) : {precision:.4f}")

print(f"Recall (Macro)    : {recall:.4f}")

print(f"F1-Score (Macro)  : {f1:.4f}")


# ============================================================
# 25. VISUALISASI HASIL PREDIKSI DATA TESTING
# ============================================================

jumlah_tampil = min(12, len(df_test))

sample_test = df_test.sample(

    n=jumlah_tampil,

    random_state=SEED

).reset_index(drop=True)

plt.figure(figsize=(14, 12))

for i, row in sample_test.iterrows():

    image = decode_image_pil(

        Path(row["path"]).read_bytes()

    )

    probability = model.predict(

        np.expand_dims(image, axis=0),

        verbose=0

    )[0]

    pred_label = int(np.argmax(probability))

    actual_label = int(row["label"])

    confidence = float(
        probability[pred_label]
    )

    warna = (

        "green"

        if pred_label == actual_label

        else "red"

    )

    plt.subplot(3, 4, i + 1)

    plt.imshow(image.astype("uint8"))

    plt.title(

        f"Aktual: {CLASS_NAMES[actual_label]}\n"

        f"Prediksi: {CLASS_NAMES[pred_label]}\n"

        f"Confidence: {confidence*100:.2f}%",

        color=warna,

        fontsize=9

    )

    plt.axis("off")

plt.tight_layout()

plt.show()


# ============================================================
# 26. FUNGSI PREDIKSI GAMBAR BARU
# ============================================================

def predict_image(image_path):

    image_path = Path(image_path)

    if not image_path.exists():

        raise FileNotFoundError(

            f"Gambar tidak ditemukan: {image_path}"

        )

    # Membaca gambar RGB

    image = decode_image_pil(

        image_path.read_bytes()

    )

    # Menambahkan batch dimension

    image_input = np.expand_dims(

        image,

        axis=0

    )

    # Melakukan prediksi

    prediction = model.predict(

        image_input,

        verbose=0

    )[0]

    # Mengambil kelas dengan probabilitas tertinggi

    pred_label = int(np.argmax(prediction))

    hasil = CLASS_NAMES[pred_label]

    confidence = float(prediction[pred_label])

    # Menampilkan gambar

    plt.figure(figsize=(6, 6))

    plt.imshow(image.astype("uint8"))

    plt.title(

        f"Hasil Prediksi: {hasil}\n"

        f"Confidence: {confidence*100:.2f}%"

    )

    plt.axis("off")

    plt.show()

    print("\n===================================")
    print("HASIL PREDIKSI")
    print("===================================")

    print(f"Prediksi   : {hasil}")

    print(
        f"Confidence : {confidence*100:.2f}%"
    )

    print("\nProbabilitas setiap kelas:")

    for i, class_name in enumerate(CLASS_NAMES):

        print(

            f"{class_name:10s}: "

            f"{prediction[i]*100:.2f}%"

        )

    return hasil


# ============================================================
# 27. PENGUJIAN GAMBAR BARU
# ============================================================

# Ganti lokasi berikut dengan foto yang ingin diuji.
#
# Contoh:
# FOTO = r"D:\foto_rempah\jahe.jpg"

FOTO = None

if FOTO is not None:

    predict_image(FOTO)


# ============================================================
# 28. INFORMASI MODEL
# ============================================================

print("\n===================================")
print("SELURUH PROSES SELESAI")
print("===================================")

print("Arsitektur : EfficientNetB0")

print("Jumlah kelas:", NUM_CLASSES)

print("Kelas:", CLASS_NAMES)

print("\nModel tersimpan di:")

print(
    (OUTPUT_DIR / "model_rimpang.keras").resolve()
)

print("\nNama kelas tersimpan di:")

print(
    (OUTPUT_DIR / "class_names.json").resolve()
)
```
